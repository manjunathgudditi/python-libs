### why numpy?
* python lists are slow
* numpy array obj is 50x faster than lists
* array obj is called **ndarray**
  * has so many methods
* Arrays are very frequently used in data science, where speed and resources are very important.

> **Data Science**: is a branch of computer science where we study **how to store, use and analyze data for deriving information from it**.

### Why is NumPy Faster?
*  arrays are stored at one continuous place in memory
  * so processes can access and manipulate them very efficiently
  * this behavior is called locality of reference in computer science

### Which Language is NumPy written in?
* written partially in Python
* but most of the parts that require fast computation are written in C or C++.

## 4. Creating Arrays

### create ndarray Object
* done using the **array()** function.

```
type(arr)
```

we can pass a list, tuple or any array-like object into the array() method,

```
import numpy as np
arr = np.array((1, 2, 3))
print(arr)
print(type(arr))
```

### Dimensions:
* A dimension in arrays is one level of array depth
> nested array: array that have arrays as their elements.

### 0-D Arrays:
* scalars
* elements


### 1-D Arrays:
* array that has 0-D arrays as its elements

### 2-D Arrays:
* array that has 1-D arrays as its elements
* often used to represent matrix or 2nd order tensors
> NumPy has a whole sub module dedicated towards matrix operations called **numpy.mat**

### 3-D arrays:
* array that has 2-D arrays (matrices) as its elements
* often used to represent a 3rd order tensor.


### Check Number of Dimensions
* **ndim** attribute



```
import numpy as np

arr = np.array(24)
arr_1 = np.array([1, 2, 3])
arr_2 = np.array([[1,2], [3,4]])
arr_3 = np.array([[[1,2], [3,4]], [[5,6], [7,8]]])

print(arr.ndim)
print(arr)

print(arr_1.ndim)
print(arr_1)

print(arr_2.ndim)
print(arr_2)

print(arr_3.ndim)
print(arr_3)

```

### Higher Dimensional Arrays
* array can have any number of dimensions.
* When the array is created, you can define the number of dimensions by using the ndmin argument.



```
import numpy as np

arr = np.array([1,2,3], ndmin=5)

print(arr)

print(arr.ndim)
```
